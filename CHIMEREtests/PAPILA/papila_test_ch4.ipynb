```python
# import libraries
import xarray as xr
import geopandas as gpd
```

```python
# open 2020 ch4 dataset
nox_data = xr.open_dataset('PAPILA_0.1x0.1_anthro_nox_v2.0_yearly_2020.nc')

# open shapefile
geopd = gpd.read_file('PozosSecEnrg.geojson')
```

```python
geopd[geopd.FECHA_F_01 > '2020-01-01']
```

```python
import matplotlib.pyplot as plt

xmin, xmax = -71.257, -66.368
ymin, ymax = -40.171, -36.319

nox_data_sel = nox_data.sel(lat=slice(ymin, ymax), lon=slice(xmin, xmax))

fig, ax = plt.subplots(1, 2, figsize=(18, 10), sharey=True)

nox_data_sel.fef.plot(ax=ax[0])#vmin=0.01, vmax=0.027)
geopd[geopd.FECHA_F_01 > '2020-01-01'].plot(ax=ax[1],
    marker='o',
    facecolor='none',
    edgecolor='red',
    markersize=50,
    linewidth=1,
    alpha=0.2)

ax[1].set_xlim(xmin, xmax)
ax[1].set_ylim(ymin, ymax)
```
