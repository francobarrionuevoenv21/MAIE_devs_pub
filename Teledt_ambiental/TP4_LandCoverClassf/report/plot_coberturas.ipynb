```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
list_clases = ['Pastizal', 'Arbustal', 'Bosque', 'Cultivo', 'Urbanización', 'Agua', 'Roca']
array_cov = np.array([26.84, 25.69, 16.25, 14.42, 12.21, 3.14, 1.44])
```

```python
fig, ax = plt.subplots(figsize=(8, 4))

colors = [ 
'#98ff00', # Arbustal // verde claro 
'#f5ff00', # Pastizal // amarillo 
'#006a06', # Bosque// verde oscuro
'#bf04c2', # Cultivo  // magenta
'#958f87', #Urbanizacion // gris
'#032cd6', # Agua  // azul 
'#ffffff', # Roca  // blanco 
]

plt.bar(list_clases, array_cov, color=colors, edgecolor="black", linewidth=1)
ax.set_title('Distribución porcentual coberturas según modelo RF 2')
ax.set_ylabel('Porcentaje ocupado (%)', )
ax.set_xlabel('Tipo de cobertura')
```
