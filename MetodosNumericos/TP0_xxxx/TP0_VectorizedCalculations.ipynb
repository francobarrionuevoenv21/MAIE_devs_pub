```python
# Define domain values
x = np.linspace(-10, 11, 100)

# Plotting
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# 5*sen(2x)
ax[0].plot(x, 5*np.sin(2*x))
ax[0].set_title('5*sen(2x)')
ax[0].set_xlabel('x')
ax[0].set_ylabel('y')

# 3*log(x^2)
ax[1].plot(x, 3*np.log(x**2))
ax[1].set_title('3*log(x^2)')
ax[1].set_xlabel('x')
ax[1].set_ylabel('y')
```

```python
# Define domain values
x = np.linspace(-5, 5, 500)

# Plotting
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# f(x) = -sen(x) cos(x^2)
ax[0].plot(x, -np.sin(x) * np.cos(x**2))
ax[0].set_title('f(x) = -sen(x) cos(x^2)')
ax[0].set_xlabel('x')
ax[0].set_ylabel('y')

# f(x) = exp(-x/2) sen^2(x)
ax[1].plot(x, np.exp(-x/2) * np.sin(x)**2)
ax[1].set_title('f(x) = exp(-x/2) sen^2(x)')
ax[1].set_xlabel('x')
ax[1].set_ylabel('y')
```

```python
9# Define domain values
x = np.linspace(0, 4, 100)

# Define functions
f1 = np.poly1d([1, -6, -12, -8]) # x^3 - 6x^2 - 12x - 8
f2 = np.poly1d([1, -8, 20, -16]) # x^3 - 8x^2 + 20x - 16
f3 = np.poly1d([1, -10, 35, -50]) # x^3 - 10x^2 + 35x - 50
f4 = np.poly1d([1, -2]) # x - 2

# Plotting
fig, ax = plt.subplots(2, 2, figsize=(8, 6), constrained_layout=True)

ax[0, 0].plot(x, f1(x))
ax[0, 0].set_title('f1(x) = x^3 - 6x^2 - 12x - 8', fontsize=10) 
ax[0, 0].set_xlabel('x')
ax[0, 0].set_ylabel('f1(x)')    


ax[0, 1].plot(x, f2(x))
ax[0, 1].set_title('f2(x) = x^3 - 8x^2 + 20x - 16', fontsize=10) 
ax[0, 1].set_xlabel('x')
ax[0, 1].set_ylabel('f2(x)')

ax[1, 0].plot(x, f3(x))
ax[1, 0].set_title('f3(x) = x^3 - 10x^2 + 35x - 50', fontsize=10)
ax[1, 0].set_xlabel('x')
ax[1, 0].set_ylabel('f3(x)')

ax[1, 1].plot(x, f4(x))
ax[1, 1].set_title('f4(x) = x - 2', fontsize=10)
ax[1, 1].set_xlabel('x')
ax[1, 1].set_ylabel('f4(x)')
```

```python
# Plotting
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# f1(x) - 2*f2(x)
ax[0].plot(x, f1(x)-2*f2(x))
ax[0].set_title('f1(x) - 2*f2(x)', fontsize=10)
ax[0].set_xlabel('x')
ax[0].set_ylabel('y')


# 3*f4(x) - f1(x) - 2*f2(x)
ax[1].plot(x, 3*f4(x)-f1(x)-2*f2(x))
ax[1].set_title('3*f4(x) - f1(x) - 2*f2(x)', fontsize=10)
ax[1].set_xlabel('x')
ax[1].set_ylabel('y')
```

```python
# Define coefficients for g1(x) = x^3 - 5x^2 + 2x + 8
g1_coeff = [1, -5, 2, 8]
g1_roots = np.roots(g1_coeff)
x_g1 = np.linspace(min(g1_roots)-1, max(g1_roots)+1, 100)

# Plot g1(x)
plt.plot(x_g1, np.polyval(g1_coeff, x_g1))
plt.axhline(0, color='gray', lw=0.5, ls='--')
plt.plot(g1_roots, np.zeros_like(g1_roots), 'ro', label='Roots')
plt.title('Plot of g1(x) = x^3 - 5x^2 + 2x + 8')
plt.xlabel('x')
plt.ylabel('g1(x)')
```
