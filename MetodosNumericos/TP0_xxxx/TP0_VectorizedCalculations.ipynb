```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.array([-2, 3, 6])
b = np.array([0, -1, 5])
```

```python
# a)
print(a-b^2)
# c)
print(4*a+3+b)
```

```python
# Define domain values
x = np.linspace(-10, 11, 100)

# Plotting
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# 5*sen(2x)
ax[0].plot(x, 5*np.sin(2*x))
ax[0].set_title('5*sen(2x)')
ax[0].set_xlabel('x')
ax[0].set_ylabel('y')

# 3*log(x^2)
ax[1].plot(x, 3*np.log(x**2))
ax[1].set_title('3*log(x^2)')
ax[1].set_xlabel('x')
ax[1].set_ylabel('y')
```

```python
# Define domain values
x = np.linspace(-5, 5, 500)

# Plotting
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# f(x) = -sen(x) cos(x^2)
ax[0].plot(x, -np.sin(x) * np.cos(x**2))
ax[0].set_title('f(x) = -sen(x) cos(x^2)')
ax[0].set_xlabel('x')
ax[0].set_ylabel('y')

# f(x) = exp(-x/2) sen^2(x)
ax[1].plot(x, np.exp(-x/2) * np.sin(x)**2)
ax[1].set_title('f(x) = exp(-x/2) sen^2(x)')
ax[1].set_xlabel('x')
ax[1].set_ylabel('y')
```
