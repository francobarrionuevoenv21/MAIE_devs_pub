```python
# Define the function to calculate x raised to the power of n
def pot_ene(x, n):

    if n == int(n): # check n integer
        return x**n
    else: # n is odd
        print('n debe ser entero')

# Define x and n for testing the function
x = 4
n = 3.1

# Test the function with defined x and n
print('Potencia x y n definidos: ')
pot_ene(x, n)

print('---------------------------------------------')

# Test function with the the first 5 natural numbers
print('Primeras cinco potencias naturales de x:')
for i in range(5):
    print(pot_ene(x, i))
```

```python
# Define the function to calculate the approximation of pi using Wallis product
def pi_aprox(n):

    # Define the domain for the function
    x_values = np.linspace(1, n, 1000)

    # Define the acumlator (prod)
    prod = 1

    # Implement a loop to compute Wallis product
    for x in x_values:
        prod *= (4 * x**2) / (4 * x**2 - 1)

    return 2 * prod

# Print the approximation of pi
print('Aproximación de pi: ', pi_aprox(1000))
```

```python
# Define n values arrays
n_values = np.arange(10, 1000, 10)

# Define an array to store the approximations of pi
pi_approximations = [pi_aprox(n) for n in n_values]

# Plot the approximations of pi against n
plt.figure(figsize=(5, 3))
plt.plot(n_values, pi_approximations, label='Aproximación de pi', color='blue')
plt.axhline(y=np.pi, color='red', linestyle='--', label='Valor real de pi')
plt.title('Aproximación de pi usando el producto de Wallis')
plt.xlabel('n')
plt.ylabel('Aproximación de pi')
plt.yscale('log')
plt.legend()
plt.show()
```

```python
# Define lists with the number of days per month (leap and non-leap years)
list_days = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
list_daysb = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

# Function to check if a year is leap or not
def leap_year(year):
    if (year % 4 == 0 and year % 100 != 0):
        return True
    elif (year % 100 == 0 and year % 400 == 0):
        return True
    else:
        return False

# Function to calculate the day of the year
def doy(day, month, year):
    if leap_year(year):
        doy = 0
        for d in list_daysb[:month-1]:
            doy += d
        return doy + day
    else:
        doy = 0
        for d in list_days[:month-1]:
            doy += d
        return doy + day

# Ask the user for the day, month and year
day = 25
month = 10
year = 2026

# Compute doy
print('Día del año: ', doy(day, month, year))
```
