##  Ejercicio 1

image.png

```python
# Creamos un objeto llamado G que es un grafo
G1 = nx.DiGraph()

# Definimos los nodos que va a contener
nodos = [
    "A","B","C","D","E","F","G"
]

# Agregamos los nodos al grafo
G1.add_nodes_from(nodos)

# Determinamos las conexiones entre nodos
aristas = [
    ("A","B", 1), ("B", "C", 0), ("C", "A", 1),
    ("C", "D", 1), ("D", "C", 1), ("D", "E", 1),
    ("F", "G", 1), ("G", "A", 1), ("F", "A", 1), ("E", "A", 1)
]

# Se agregan las conexiones al grafo
G1.add_weighted_edges_from(aristas) # Tambien se puede con add_edged_from(), y se puede agregar un diccionario con features
```

```python
# Se define las características de la figura a dibujar
plt.figure(figsize=(6,6))
pos = nx.spring_layout(G1, k=2, seed=42)

nx.draw(
    G1, pos,
    with_labels=True,
    node_color="lightblue",
    node_size=2000/2,
    font_size=10,
)

# Dibujamos el grafo
plt.show()
```

```python
# Creamos un objeto llamado G que es un grafo
G2 = nx.Graph()

# Definimos los nodos que va a contener
nodos = [
    "a", "b", "c", "d", "e"
]

# Agregamos los nodos al grafo
G2.add_nodes_from(nodos)

# Determinamos las conexiones entre nodos
aristas = [
    ("a", "b"), ("b", "c"), ("b", "e"), ("e", "d"), ("e", "c"), ("a", "d")
]

# Se agregan las conexiones al grafo
G2.add_edges_from(aristas)
```

```python
# Se define las características de la figura a dibujar
plt.figure(figsize=(6,6))
pos = nx.spring_layout(G2, seed=42)

nx.draw(
    G2, pos,
    with_labels=True,
    node_color="lightblue",
    node_size=2000,
    font_size=10,
)

# Dibujamos el grafo
plt.show()
```

```python
nodos = [
    "ny", "ch", "to", "de", "la", "ho"
]

aristas = [
    ("ny","ch", 1000), ("ny","to", 2000), ("ny", "de", 1800),
    ("to", "ch", 700), ("to", "la", 1000),
    ("ch", "de", 600),
    ("de", "la", 1500), ("de", "ho", 500),
    ("ho", "la", 800)
]

# Creamos un objeto llamado G que es un grafo
G4 = nx.DiGraph()

# Agregamos los nodos al grafo
G4.add_nodes_from(nodos)

# Se agregan las conexiones al grafo
G4.add_weighted_edges_from(aristas)
```

```python
# Se define las características de la figura a dibujar
plt.figure(figsize=(6,6))
pos = nx.spring_layout(G4, k=2, seed=42)

nx.draw(
    G4, pos,
    with_labels=True,
    node_color="lightblue",
    node_size=2000,
    font_size=10,
)

# Dibujamos el grafo
plt.show()
```
